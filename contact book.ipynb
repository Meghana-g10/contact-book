```python
import tkinter as tk
from tkinter import messagebox
from tkinter import ttk

# Data structure
contacts = []

# Add contact
def add_contact():
    name = entry_name.get()
    phone = entry_phone.get()
    email = entry_email.get()
    address = entry_address.get()
    pincode = entry_pincode.get()

    if not name or not phone:
        messagebox.showerror("Error", "Name and Phone are required!")
        return

    contacts.append({
        'Name': name,
        'Phone': phone,
        'Email': email,
        'Address': address,
        'Pincode': pincode
    })
    update_treeview()
    clear_entries()

# Delete contact
def delete_contact():
    selected = tree.selection()
    if not selected:
        messagebox.showwarning("Warning", "Please select a contact to delete.")
        return

    index = tree.index(selected)
    del contacts[index]
    update_treeview()
    clear_entries()

# Select contact for update
def select_contact(event):
    selected = tree.selection()
    if selected:
        index = tree.index(selected)
        contact = contacts[index]
        entry_name.delete(0, tk.END)
        entry_name.insert(0, contact['Name'])
        entry_phone.delete(0, tk.END)
        entry_phone.insert(0, contact['Phone'])
        entry_email.delete(0, tk.END)
        entry_email.insert(0, contact['Email'])
        entry_address.delete(0, tk.END)
        entry_address.insert(0, contact['Address'])
        entry_pincode.delete(0, tk.END)
        entry_pincode.insert(0, contact['Pincode'])

# Update contact
def update_contact():
    selected = tree.selection()
    if not selected:
        messagebox.showwarning("Warning", "Please select a contact to update.")
        return

    index = tree.index(selected)
    contacts[index] = {
        'Name': entry_name.get(),
        'Phone': entry_phone.get(),
        'Email': entry_email.get(),
        'Address': entry_address.get(),
        'Pincode': entry_pincode.get()
    }
    update_treeview()
    clear_entries()

# View contact in popup
def view_contact():
    selected = tree.selection()
    if not selected:
        messagebox.showwarning("Warning", "Please select a contact to view.")
        return

    index = tree.index(selected)
    contact = contacts[index]

    popup = tk.Toplevel(root)
    popup.title("Contact Details")
    popup.geometry("300x250")
    popup.config(bg="#FFF8DC")

    tk.Label(popup, text="Contact Details", font=("Arial", 14, "bold"), bg="#FFD700", fg="black", pady=5).pack(fill=tk.X)

    details = f"""
Name   : {contact['Name']}
Phone  : {contact['Phone']}
Email  : {contact['Email']}
Address: {contact['Address']}
Pincode: {contact['Pincode']}
"""
    tk.Label(popup, text=details.strip(), font=("Arial", 12), justify="left", bg="#FFF8DC").pack(padx=10, pady=10)

# Refresh table
def update_treeview():
    tree.delete(*tree.get_children())
    for contact in contacts:
        tree.insert('', 'end', values=(
            contact['Name'],
            contact['Phone'],
            contact['Email'],
            contact['Address'],
            contact['Pincode']
        ))

# Clear form
def clear_entries():
    entry_name.delete(0, tk.END)
    entry_phone.delete(0, tk.END)
    entry_email.delete(0, tk.END)
    entry_address.delete(0, tk.END)
    entry_pincode.delete(0, tk.END)

# GUI Setup
root = tk.Tk()
root.title("Colorful Contact Book")
root.geometry("800x600")
root.config(bg="#F5F5F5")

# Header
header = tk.Label(root, text="📒 My Contact Book", font=("Helvetica", 20, "bold"), bg="#00ADB5", fg="white", pady=10)
header.pack(fill=tk.X)

# Entry Frame
frame = tk.Frame(root, bg="#F5F5F5")
frame.place(x=30, y=60)

tk.Label(frame, text="Name:", font=("Arial", 12), bg="#F5F5F5").grid(row=0, column=0, sticky="e", pady=5)
entry_name = tk.Entry(frame, width=35, font=("Arial", 12))
entry_name.grid(row=0, column=1)

tk.Label(frame, text="Phone:", font=("Arial", 12), bg="#F5F5F5").grid(row=1, column=0, sticky="e", pady=5)
entry_phone = tk.Entry(frame, width=35, font=("Arial", 12))
entry_phone.grid(row=1, column=1)

tk.Label(frame, text="Email:", font=("Arial", 12), bg="#F5F5F5").grid(row=2, column=0, sticky="e", pady=5)
entry_email = tk.Entry(frame, width=35, font=("Arial", 12))
entry_email.grid(row=2, column=1)

tk.Label(frame, text="Address:", font=("Arial", 12), bg="#F5F5F5").grid(row=3, column=0, sticky="e", pady=5)
entry_address = tk.Entry(frame, width=35, font=("Arial", 12))
entry_address.grid(row=3, column=1)

tk.Label(frame, text="Pincode:", font=("Arial", 12), bg="#F5F5F5").grid(row=4, column=0, sticky="e", pady=5)
entry_pincode = tk.Entry(frame, width=35, font=("Arial", 12))
entry_pincode.grid(row=4, column=1)

# Buttons Frame
btn_frame = tk.Frame(root, bg="#F5F5F5")
btn_frame.place(x=450, y=60)

tk.Button(btn_frame, text="Add Contact", command=add_contact, width=20, bg="#00C897", fg="white", font=("Arial", 11, "bold")).pack(pady=5)
tk.Button(btn_frame, text="Update Contact", command=update_contact, width=20, bg="#F7C04A", fg="black", font=("Arial", 11, "bold")).pack(pady=5)
tk.Button(btn_frame, text="Delete Selected", command=delete_contact, width=20, bg="#FF6464", fg="white", font=("Arial", 11, "bold")).pack(pady=5)
tk.Button(btn_frame, text="View Contact", command=view_contact, width=20, bg="#7AA2E3", fg="white", font=("Arial", 11, "bold")).pack(pady=5)

# Table Styling
style = ttk.Style()
style.theme_use("clam")
style.configure("Treeview.Heading", font=("Arial", 12, "bold"), background="#393E46", foreground="white")
style.configure("Treeview", font=("Arial", 11), rowheight=25, background="#EEEEEE", fieldbackground="#EEEEEE")

# Contact Table
tree = ttk.Treeview(root, columns=("Name", "Phone", "Email", "Address", "Pincode"), show='headings')
for col in ("Name", "Phone", "Email", "Address", "Pincode"):
    tree.heading(col, text=col)
    tree.column(col, anchor="center", width=140)

tree.place(x=30, y=300, width=740, height=250)
tree.bind("<<TreeviewSelect>>", select_contact)

# Run App
root.mainloop()
```
